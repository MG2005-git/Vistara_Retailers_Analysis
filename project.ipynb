### Data Preprocessing

```python
# checking if the dataset has any duplicate values

df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
#checking if the dataset has any null values

df.isnull().sum()
```

```python
# dropping the all the rows with null values from the dataset

df = df.dropna()
```

```python
df.shape
```

```python
df.dtypes
```

### Feature Engineering

```python
#Above we can see that unitprice has datatype as str so we need to convert it into float so that we can perform the analysis on it.

# as we have $ sign in the unitprice column we need to remove it and then convert the datatype into float.

df['unit_price'] = df['unit_price'].str.replace('$', '').astype(float)
```

```python
df.info()
```

```python
# Changing the datatype of date column from object to datetime

df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y') 
```

```python
df.columns
```

```python
#adding the total price column to the dataset by multiplying the quantity and unit price columns

df['total_price'] = df['quantity'] * df['unit_price']
```

```python
df.head()
```

```python
# adding a profit column by multiplying the total price column with profit_margin column

df['profit'] = df['total_price'] * df['profit_margin']
```

```python
df.shape
```

```python
# downloading the cleaned dataset 

df.to_csv('cleaned_vistaradata.csv', index=False)
```
